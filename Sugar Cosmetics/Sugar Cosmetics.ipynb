```python
lis = ['Comments', 'Saves', 'Impressions', 'Reach', 'Shares', 'Profile Visits', 'Follows']
for i in lis:
    k1 = i+"_statistic"
    k2 = i+"_p_value"
    v1, v2 = scipy.stats.kstest(df[i], 'norm')
    print(f"{k1}: {v1}, {k2}:{v2}")
    
```
